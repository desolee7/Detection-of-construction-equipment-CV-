```python
import json
from pathlib import Path
from PIL import Image

# ============================================================
# PATHS
# ============================================================

DART_ROOT = Path(r"DART\Liebherr_Product")
LABELS_PATH = DART_ROOT / "labels" / "labels_nms.json"

OUTPUT_ROOT = Path(r"Liebherr_Product_YOLO")
IMAGES_DIR = OUTPUT_ROOT / "images"
LABELS_DIR = OUTPUT_ROOT / "labels"

IMAGES_DIR.mkdir(parents=True, exist_ok=True)
LABELS_DIR.mkdir(parents=True, exist_ok=True)

# ============================================================
# LOAD DART LABELS
# ============================================================

with open(LABELS_PATH, "r", encoding="utf-8") as f:
    labels_nms = json.load(f)

print(f"DART annotations: {len(labels_nms)}")

# ============================================================
# CLASS MAPPING
# ============================================================

CLASS_NAMES = ds["train"].features["label"].names
CLASS_TO_ID = {name: i for i, name in enumerate(CLASS_NAMES)}

print(f"Classes: {len(CLASS_NAMES)}")

# ============================================================
# CONVERSION
# ============================================================

converted = 0
skipped = 0
total_boxes = 0

for image_id, annotation in labels_nms.items():

    # --------------------------------------------------------
    # Find corresponding image in HF dataset
    # --------------------------------------------------------
    #
    # DART IDs are zero-padded:
    # 00001, 00002, ...
    #
    # HF dataset contains the images in the same dataset.
    # We use numeric ID -> dataset row.
    #
    # --------------------------------------------------------

    row_idx = int(image_id) - 1

    if row_idx < 0 or row_idx >= len(ds["train"]):
        skipped += 1
        continue

    sample = ds["train"][row_idx]
    image = sample["image"]

    # --------------------------------------------------------
    # Save image
    # --------------------------------------------------------

    image_filename = f"{image_id}.jpg"
    image_path = IMAGES_DIR / image_filename

    if image.mode != "RGB":
        image = image.convert("RGB")

    image.save(image_path, quality=95)

    # --------------------------------------------------------
    # YOLO label
    # --------------------------------------------------------

    boxes = annotation.get("boxes", [])
    classes = annotation.get("classes", [])

    yolo_lines = []

    for box, class_name in zip(boxes, classes):

        if class_name not in CLASS_TO_ID:
            print(
                f"WARNING: unknown class '{class_name}' "
                f"in image {image_id}"
            )
            continue

        class_id = CLASS_TO_ID[class_name]

        # DART bbox:
        # [x_center, y_center, width, height]
        #
        # Values are already normalized to [0, 1].
        x_center, y_center, width, height = box

        yolo_lines.append(
            f"{class_id} "
            f"{x_center:.6f} "
            f"{y_center:.6f} "
            f"{width:.6f} "
            f"{height:.6f}"
        )

    # --------------------------------------------------------
    # Save .txt
    # --------------------------------------------------------

    label_path = LABELS_DIR / f"{image_id}.txt"

    with open(label_path, "w", encoding="utf-8") as f:
        f.write("\n".join(yolo_lines))

    converted += 1
    total_boxes += len(yolo_lines)

    # Progress
    if converted % 500 == 0:
        print(
            f"Converted: {converted}/{len(labels_nms)} "
            f"| boxes: {total_boxes}"
        )

# ============================================================
# DATASET YAML
# ============================================================

yaml_path = OUTPUT_ROOT / "dataset.yaml"

with open(yaml_path, "w", encoding="utf-8") as f:

    f.write(f"path: {OUTPUT_ROOT.resolve()}\n")
    f.write("train: images\n")
    f.write("val: images\n")
    f.write("test: images\n")
    f.write(f"nc: {len(CLASS_NAMES)}\n")
    f.write("names:\n")

    for i, name in enumerate(CLASS_NAMES):
        f.write(f"  {i}: '{name}'\n")

# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("CONVERSION COMPLETE")
print("=" * 60)

print(f"Source images:       {len(ds['train'])}")
print(f"DART annotations:    {len(labels_nms)}")
print(f"Converted images:    {converted}")
print(f"Skipped:             {skipped}")
print(f"Total bounding boxes:{total_boxes}")
print(f"Classes:             {len(CLASS_NAMES)}")

print(f"\nOutput:")
print(OUTPUT_ROOT.resolve())
```
